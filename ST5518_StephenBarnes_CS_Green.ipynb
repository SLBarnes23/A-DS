```python
"""
import hashlib


# Step 1: Prompt user to enter password

password = input("Enter your password: ")

# Step 2: Hash the password using SHA-256

hashed_password = hashlib.sha256(password.encode()).hexdigest()

# Step 3: Store the hashed password in a file

file_path = "hashed_passwords.txt"
with open(file_path, "a") as file:
    file.write(hashed_password + "\n")

# Read and display the contents of the hashed_passwords.txt file

with open("hashed_passwords.txt", "r") as file:
  content = file.read()
  print("Stored Hashes:\n", content)
"""
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad
import os

#Step 1: Prompt the user for a message

message = input("Enter a message to encrypt: ")
message_bytes = message.encode()

#Step 2: Create  key and IV

key = get_random_bytes(16) # AES-128 (16 bytes)
iv = get_random_bytes(16) # CBC requires a random IV

# Step 3: Encrypt using AES CBC
cipher_encrypt = AES.new(key, AES.MODE_CBC, iv)
ciphertext = cipher_encrypt.encrypt(pad(message_bytes, AES.block_size))

# 🟡 Show encrypted message (as hex for readability)
print("Encrypted message (hex):", ciphertext.hex())

# Step 4: Save Ciphertext to file
with open("encrypted_message.bin", "wb") as f:
    f.write(iv + ciphertext)  # Save IV + ciphertext together

print("Message encrypted and saved to encrypted_message.bin")

# Step 5: Read from file and decrypt
with open("encrypted_message.bin", "rb") as f:
    file_data = f.read()
    iv_from_file = file_data[:16]
    ciphertext_from_file = file_data[16:]

cipher_decrypt = AES.new(key, AES.MODE_CBC, iv_from_file)
plaintext_padded = cipher_decrypt.decrypt(ciphertext_from_file)
plaintext = unpad(plaintext_padded, AES.block_size).decode()

# 🔵 Show decrypted message
print("Decrypted message:", plaintext)


```
